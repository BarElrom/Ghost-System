```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ---- Load CSV ----
df = pd.read_csv("example_csi.csv", header=None)

# column indices (ESP32 CSI format)
TIME_COL = 23
CSI_COL = 25

# ---- Parse CSI ----
def parse_csi(csi_str):
    csi = np.fromstring(csi_str.strip("[]"), sep=",")
    I = csi[0::2]
    Q = csi[1::2]
    return np.abs(I + 1j * Q)  # amplitude

amplitudes = np.vstack(df[CSI_COL].apply(parse_csi))
time = df[TIME_COL].values

# ---- Build axes ----
T, S = np.meshgrid(time, np.arange(amplitudes.shape[1]))

# ---- Plot 3D ----
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(
    T,
    S,
    amplitudes.T,
    cmap="viridis",
    linewidth=0,
    antialiased=True
)

ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Subcarrier Index")
ax.set_zlabel("Amplitude")

plt.tight_layout()
plt.show()
```
